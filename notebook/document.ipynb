```python
from langchain_core.documents import Document
```

```python

# Read the text file
with open(r"C:\Users\subas\RAG-PROJECT\data\alice_in_wonderland.md", "r", encoding="utf-8") as f:
    text_content = f.read()

# Wrap it directly into the core LangChain Document format
docs = [Document(page_content=text_content, metadata={"source": r"C:\Users\subas\RAG-PROJECT\data\alice_in_wonderland.md"})]
```
